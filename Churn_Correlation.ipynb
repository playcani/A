# Churn Correlation + Cleaning

**Problem:** For the Orange Telecom Churn dataset, perform EDA, calculate the correlation matrix and visualize it using a heatmap. Clean the given dataset.

```python
# Program: EDA, data cleaning and correlation heatmap for the Orange Telecom Churn dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the data
df = pd.read_csv("../ref/Orange_Telecom_Churn_Data.csv")
print("Shape:", df.shape)
df.head()
```

```python
# ---------- EDA ----------
df.info()
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Summary statistics and how many customers churned
print(df["churned"].value_counts())
df.describe()
```

```python
# ---------- CLEANING ----------
# 1. Remove duplicate rows
df = df.drop_duplicates()

# 2. Fill missing values: numbers -> median, text -> most common value
for col in df.columns:
    if df[col].dtype == "object" or df[col].dtype == "str":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())

# 3. Drop columns that are just IDs / not useful for analysis
df = df.drop(columns=["state", "area_code", "phone_number"], errors="ignore")

# 4. Convert yes/no and True/False columns into 1/0
df["intl_plan"] = df["intl_plan"].map({"yes": 1, "no": 0})
df["voice_mail_plan"] = df["voice_mail_plan"].map({"yes": 1, "no": 0})
df["churned"] = df["churned"].astype(str).str.strip().map({"True": 1, "False": 0})

print("Shape after cleaning:", df.shape)
df.head()
```

```python
# ---------- CORRELATION MATRIX ----------
corr = df.corr()

# Which features are most related to churn?
corr["churned"].sort_values(ascending=False)
```

```python
# ---------- HEATMAP ----------
plt.figure(figsize=(14, 10))
sns.heatmap(corr, annot=True, fmt=".1f", cmap="coolwarm")
plt.title("Correlation Heatmap - Telecom Churn")
plt.show()
```
