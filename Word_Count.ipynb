# Word Count

**Problem:** Write a Python program to read a text file and count the number of words in it.

```python
# Program: Read a text file and count the number of words in it

# First, create a sample text file (skip this cell if you already have your own file)
with open("sample.txt", "w") as f:
    f.write("Python is easy to learn.\n")
    f.write("Data science uses Python a lot.\n")
    f.write("Practice makes you better at Python.\n")
print("sample.txt created")
```

```python
# Read the file and count words
file_name = "sample.txt"

with open(file_name, "r") as f:
    text = f.read()

words = text.split()          # split() breaks the text wherever there is a space or new line

print("File content:\n")
print(text)
print("Total number of words:", len(words))
```

```python
# Bonus: count how many times each word appears
from collections import Counter

clean_words = [w.lower().strip(".,!?") for w in words]
print(Counter(clean_words).most_common(5))
```
