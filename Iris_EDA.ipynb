# Iris EDA

**Problem:** Use the Iris dataset. Perform basic exploratory data analysis (EDA) and visualize the findings using Matplotlib.

```python
# Program: Basic EDA on the Iris flower dataset with Matplotlib charts

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the data
df = pd.read_csv("../ref/iris.csv")
df.head()
```

```python
# Basic information about the data
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nMissing values:\n", df.isnull().sum())
df.info()
```

```python
# Summary statistics
df.describe()
```

```python
# How many flowers of each species?
print(df["species"].value_counts())

# Average measurements for each species
df.groupby("species").mean()
```

```python
# Chart 1: Bar chart - count of each species
df["species"].value_counts().plot(kind="bar", color=["skyblue", "orange", "green"])
plt.title("Count of Each Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
# Chart 2: Histograms of all 4 measurements
df.hist(figsize=(8, 6), bins=15, edgecolor="black")
plt.suptitle("Distribution of Features")
plt.show()
```

```python
# Chart 3: Scatter plot - petal length vs petal width, coloured by species
colors = {"Setosa": "red", "Versicolor": "green", "Virginica": "blue"}
for sp, group in df.groupby("species"):
    plt.scatter(group["petal.length"], group["petal.width"], label=sp, color=colors.get(sp))
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()
plt.show()
```

```python
# Chart 4: Box plot of all features (helps spot spread and outliers)
df.drop(columns="species").plot(kind="box", figsize=(8, 5))
plt.title("Box Plot of Features")
plt.show()
```

```python
# Findings:
# 1. There are 150 rows, 50 for each species, and no missing values.
# 2. Setosa has clearly smaller petals than the other two species.
# 3. Petal length and petal width grow together (strong positive relation).
# 4. Sepal width has a few outliers (seen in the box plot).
```
