# Z-Score Outliers

**Problem:** Use Python to identify outliers using the Z-score method and remove the outliers and display the cleaned DataFrame. (Using insurance.csv - just change the file name to use any other dataset.)

```python
# Program: Find and remove outliers using the Z-score method (on the insurance dataset)

import pandas as pd
```

```python
# Load the data (change the file name here to use any other dataset)
df = pd.read_csv("../ref/insurance.csv")
print("Original shape:", df.shape)
df.head()
```

```python
# Take only numeric columns, because Z-score works on numbers
num_cols = df.select_dtypes(include="number").columns
print("Numeric columns:", list(num_cols))

# Z-score = (value - mean) / std
z = ((df[num_cols] - df[num_cols].mean()) / df[num_cols].std()).abs()
z.head()
```

```python
# Any value with |Z| > 3 is an outlier
outlier_rows = (z > 3).any(axis=1)
print("Number of outlier rows:", outlier_rows.sum())
df[outlier_rows]
```

```python
# Remove outliers and display the cleaned DataFrame
clean_df = df[~outlier_rows]
print("Shape before:", df.shape)
print("Shape after :", clean_df.shape)
clean_df
```
