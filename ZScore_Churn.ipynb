# Z-Score Outliers (Churn)

**Problem:** Use the Orange_Telecom_Churn_Data dataset. Use Python to identify outliers using the Z-score method, remove the outliers and display the cleaned DataFrame.

```python
# Program: Find and remove outliers in the Orange Telecom Churn data using the Z-score method

import pandas as pd
```

```python
# Load the dataset (the CSV files are kept in the ref folder)
df = pd.read_csv("../ref/Orange_Telecom_Churn_Data.csv")
print("Original shape:", df.shape)
df.head()
```

```python
# Z-score works only on numbers, so take the numeric columns
num_cols = df.select_dtypes(include="number").columns
print("Numeric columns:", list(num_cols))

# Z-score = (value - mean) / std  -> tells how far a value is from the average
z = ((df[num_cols] - df[num_cols].mean()) / df[num_cols].std()).abs()
z.head()
```

```python
# A value with |Z| > 3 is treated as an outlier
outlier_rows = (z > 3).any(axis=1)
print("Rows having at least one outlier:", outlier_rows.sum())
df[outlier_rows].head()
```

```python
# Remove the outlier rows and display the cleaned DataFrame
clean_df = df[~outlier_rows]
print("Shape before:", df.shape)
print("Shape after :", clean_df.shape)
clean_df
```
