# Housing EDA

**Problem:** Use the housing price dataset. Perform basic exploratory data analysis (EDA) and visualize the findings using Matplotlib.

```python
# Program: Basic EDA on the housing price dataset with Matplotlib charts

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the data
df = pd.read_csv("../ref/housing.csv")
df.head()
```

```python
# Basic information
print("Shape:", df.shape)
df.info()
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Summary statistics
df.describe()
```

```python
# Count of houses in each ocean_proximity category, and their average price
print(df["ocean_proximity"].value_counts())
df.groupby("ocean_proximity")["median_house_value"].mean()
```

```python
# Chart 1: Histogram of house prices
plt.hist(df["median_house_value"], bins=40, color="skyblue", edgecolor="black")
plt.title("Distribution of House Prices")
plt.xlabel("Median House Value")
plt.ylabel("Number of Areas")
plt.show()
```

```python
# Chart 2: Bar chart - average price by ocean proximity
df.groupby("ocean_proximity")["median_house_value"].mean().plot(kind="bar", color="orange")
plt.title("Average House Price by Ocean Proximity")
plt.ylabel("Average Price")
plt.show()
```

```python
# Chart 3: Scatter plot - income vs house price
plt.scatter(df["median_income"], df["median_house_value"], alpha=0.2)
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.title("Income vs House Price")
plt.show()
```

```python
# Chart 4: Location map - each point is an area, colour shows price
plt.figure(figsize=(8, 6))
plt.scatter(df["longitude"], df["latitude"], c=df["median_house_value"], cmap="jet", s=5)
plt.colorbar(label="House Value")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("House Prices by Location")
plt.show()
```

```python
# Chart 5: Box plot of house age
plt.boxplot(df["housing_median_age"])
plt.title("Box Plot of Housing Median Age")
plt.show()
```

```python
# Findings:
# 1. total_bedrooms has 207 missing values; all other columns are complete.
# 2. Prices are capped at 500001, which is why the histogram has a spike at the end.
# 3. Higher income areas have higher house prices (clear upward trend).
# 4. Houses near the ocean/bay are costlier, INLAND houses are the cheapest.
```
