# Churn Seaborn Plots

**Problem:** For the Orange_Telecom_Churn_Data dataset, use Seaborn to create a histogram, box plot, and heatmap.

```python
# Program: Draw a histogram, box plot and heatmap for the Orange Telecom Churn data using Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv("../ref/Orange_Telecom_Churn_Data.csv")
print("Shape:", df.shape)
df.head()
```

```python
# 1. HISTOGRAM - how total day minutes are spread among customers
sns.histplot(data=df, x="total_day_minutes", bins=30, kde=True, color="teal")
plt.title("Histogram of Total Day Minutes")
plt.show()
```

```python
# 2. BOX PLOT - compare total day minutes for customers who churned vs stayed
sns.boxplot(data=df, x="churned", y="total_day_minutes")
plt.title("Total Day Minutes vs Churn")
plt.show()
```

```python
# 3. HEATMAP - correlation between all numeric columns
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".1f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
