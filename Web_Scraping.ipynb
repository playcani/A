# Web Scraping with BeautifulSoup

**Problem:** Identify a website with tabular data (e.g., a Wikipedia table). Use BeautifulSoup to scrape the data, store it in a Data Frame and perform any 5 operations on the data frame.

```python
# Program: Scrape the Indian states population table from Wikipedia using BeautifulSoup and do 5 operations

import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Step 1: Download the web page (User-Agent header is needed, otherwise Wikipedia may block us)
url = "https://en.wikipedia.org/wiki/List_of_states_and_union_territories_of_India_by_population"
response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
print("Status code:", response.status_code)   # 200 means success
```

```python
# Step 2: Parse the HTML and find the first table with class "wikitable"
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", class_="wikitable")

# Step 3: Go through each row and pick the columns we need
rows = []
for tr in table.find_all("tr"):
    cells = [c.get_text(strip=True) for c in tr.find_all(["th", "td"])]
    # skip heading rows and the final "Total" row
    if len(cells) > 5 and cells[0] not in ("Rank", "Total"):
        rows.append([cells[1], cells[2], cells[3]])

df = pd.DataFrame(rows, columns=["State", "Population", "Percent_of_India"])
df.head()
```

```python
# Operation 1: Shape and info of the DataFrame
print("Rows, Columns:", df.shape)
df.info()
```

```python
# Operation 2: Clean the data - remove commas / % signs and convert to numbers
df["Population"] = df["Population"].str.replace(",", "").astype(int)
df["Percent_of_India"] = df["Percent_of_India"].str.replace("%", "").astype(float)
df.head()
```

```python
# Operation 3: Sort and show the top 5 most populated states
df.sort_values(by="Population", ascending=False).head(5)
```

```python
# Operation 4: Filter - states with population more than 10 crore (100 million)
df[df["Population"] > 100_000_000]
```

```python
# Operation 5: Add a new column - population in crores, and see basic statistics
df["Population_Crore"] = (df["Population"] / 10_000_000).round(2)
print(df[["State", "Population_Crore"]].head())
df.describe()
```
