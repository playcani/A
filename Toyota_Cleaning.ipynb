# Toyota Data Cleaning

**Problem:** For the Toyota dataset, use Python to handle missing values, outliers, duplicate rows. Display the data frame after each operation.

```python
# Program: Clean the Toyota car dataset - handle missing values, outliers and duplicate rows

import pandas as pd
```

```python
# Load the data. "??" and "????" are junk values in this file, so we treat them as missing (NaN)
df = pd.read_csv("../ref/Toyota.csv", index_col=0, na_values=["??", "????"])
print("Shape:", df.shape)
df.head()
```

```python
# Look at the data types and how many values are missing in each column
df.info()
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
# ---------- 1. HANDLE MISSING VALUES ----------
# Numeric columns -> fill with mean (average)
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["KM"]  = df["KM"].fillna(df["KM"].mean())
df["HP"]  = df["HP"].fillna(df["HP"].mean())

# Category type columns -> fill with mode (most common value)
df["FuelType"] = df["FuelType"].fillna(df["FuelType"].mode()[0])
df["MetColor"] = df["MetColor"].fillna(df["MetColor"].mode()[0])

# Doors column has words like "three", "four" - change them to numbers
df["Doors"] = df["Doors"].replace({"two": "2", "three": "3", "four": "4", "five": "5"}).astype(int)

print("Missing values after filling:")
print(df.isnull().sum())
df.head()
```

```python
# ---------- 2. HANDLE OUTLIERS (IQR method on Price) ----------
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print("Lower limit:", lower, " Upper limit:", upper)

outliers = df[(df["Price"] < lower) | (df["Price"] > upper)]
print("Number of outliers found:", len(outliers))

# keep only the rows that are inside the limits
df = df[(df["Price"] >= lower) & (df["Price"] <= upper)]
print("Shape after removing outliers:", df.shape)
df.head()
```

```python
# ---------- 3. HANDLE DUPLICATE ROWS ----------
print("Duplicate rows found:", df.duplicated().sum())

df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
df.head()
```
