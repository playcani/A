# SQLite + DataFrame

**Problem:** Set up a local SQLite database and create a table with some sample data. Use Python to import data from the SQL table into a data frame and perform any 5 operations on the data frame.

```python
# Program: Create a SQLite table, load it into a pandas DataFrame and do 5 operations on it

import sqlite3
import pandas as pd
```

```python
# Step 1: Connect to a database file (it gets created automatically if it doesn't exist)
conn = sqlite3.connect("students.db")
cur = conn.cursor()

# Step 2: Create a table (drop the old one first so we can re-run this cell safely)
cur.execute("DROP TABLE IF EXISTS students")
cur.execute("""
    CREATE TABLE students (
        id    INTEGER PRIMARY KEY,
        name  TEXT,
        age   INTEGER,
        city  TEXT,
        marks INTEGER
    )
""")

# Step 3: Insert some sample rows
data = [
    (1, "Aarav",  20, "Pune",   85),
    (2, "Priya",  21, "Mumbai", 92),
    (3, "Rohan",  19, "Pune",   67),
    (4, "Sneha",  22, "Delhi",  78),
    (5, "Karan",  20, "Mumbai", 55),
    (6, "Ananya", 21, "Delhi",  88),
    (7, "Vikram", 23, "Pune",   73),
    (8, "Isha",   20, "Mumbai", 95),
]
cur.executemany("INSERT INTO students VALUES (?, ?, ?, ?, ?)", data)
conn.commit()
print("Table created and sample data inserted!")
```

```python
# Step 4: Read the SQL table into a DataFrame
df = pd.read_sql("SELECT * FROM students", conn)
df
```

```python
# Operation 1: Basic statistics of the numeric columns
df.describe()
```

```python
# Operation 2: Filter - students who scored more than 80
df[df["marks"] > 80]
```

```python
# Operation 3: Sort students by marks (highest first)
df.sort_values(by="marks", ascending=False)
```

```python
# Operation 4: Group by city and find the average marks
df.groupby("city")["marks"].mean()
```

```python
# Operation 5: Add a new column "result" - Pass if marks >= 60, else Fail
df["result"] = df["marks"].apply(lambda m: "Pass" if m >= 60 else "Fail")
df
```

```python
# Always close the connection when done
conn.close()
print("Connection closed.")
```
