# One-Sample t-Test

**Problem:** Perform a hypothesis test to determine if the mean of a sample differs significantly from a known population mean.

```python
# Program: One-sample t-test to check if a sample mean is different from a known population mean

import numpy as np
from scipy import stats
```

```python
# Example: The average weight of students in a college is known to be 60 kg.
# We measured 15 students. Is their average different from 60?

sample = [62, 58, 65, 70, 61, 59, 66, 68, 63, 64, 57, 69, 62, 67, 65]
population_mean = 60

print("Sample mean:", np.mean(sample))
print("Population mean:", population_mean)
```

```python
# Hypotheses:
#   H0 (null)        : sample mean = population mean  (no difference)
#   H1 (alternative) : sample mean != population mean (there is a difference)

t_stat, p_value = stats.ttest_1samp(sample, population_mean)

print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_value, 4))
```

```python
# Decision using significance level alpha = 0.05
alpha = 0.05

if p_value < alpha:
    print("p-value < 0.05 -> Reject H0")
    print("Conclusion: The sample mean is significantly different from the population mean.")
else:
    print("p-value >= 0.05 -> Fail to reject H0")
    print("Conclusion: No significant difference between sample mean and population mean.")
```
