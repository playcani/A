# Housing Correlation + Cleaning

**Problem:** For the housing price dataset, perform EDA, calculate the correlation matrix and visualize it using a heatmap. Clean the given dataset.

```python
# Program: EDA, data cleaning and correlation heatmap for the housing price dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the data
df = pd.read_csv("../ref/housing.csv")
print("Shape:", df.shape)
df.head()
```

```python
# ---------- EDA ----------
df.info()
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
df.describe()
```

```python
# ---------- CLEANING ----------
# 1. Fill missing total_bedrooms with the median
df["total_bedrooms"] = df["total_bedrooms"].fillna(df["total_bedrooms"].median())

# 2. Remove duplicate rows (if any)
df = df.drop_duplicates()

# 3. Convert text column ocean_proximity into numbers (one-hot encoding)
df = pd.get_dummies(df, columns=["ocean_proximity"], dtype=int)

print("Missing values left:", df.isnull().sum().sum())
print("Shape after cleaning:", df.shape)
df.head()
```

```python
# ---------- CORRELATION MATRIX ----------
corr = df.corr()
corr
```

```python
# Which features are most related to house price?
corr["median_house_value"].sort_values(ascending=False)
```

```python
# ---------- HEATMAP ----------
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap - Housing Data")
plt.show()
```

```python
# Findings:
# 1. median_income has the strongest positive correlation with house price (~0.69).
# 2. INLAND houses have a negative correlation with price (they are cheaper).
# 3. total_rooms, total_bedrooms, population and households are strongly related to each other.
```
